## Implementación y Recorridos de Árboles Binarios

```python
class Nodo:
    def __init__(self, valor):
        self.valor = valor
        self.izq = None
        self.der = None

# Utilidades de recorrido
def preorden(nodo):
    if nodo:
        print(nodo.valor, end=" | ")
        preorden(nodo.izq)
        preorden(nodo.der)

def inorden(nodo):
    if nodo:
        inorden(nodo.izq)
        print(nodo.valor, end=" | ")
        inorden(nodo.der)

def postorden(nodo):
    if nodo:
        postorden(nodo.izq)
        postorden(nodo.der)
        print(nodo.valor, end=" | ")

# Funciones de análisis
def contar_nodos(nodo):
    if nodo is None:
        return 0
    return 1 + contar_nodos(nodo.izq) + contar_nodos(nodo.der)

def contar_hojas(nodo):
    if nodo is None:
        return 0
    if nodo.izq is None and nodo.der is None:
        return 1
    return contar_hojas(nodo.izq) + contar_hojas(nodo.der)

def buscar_valor(nodo, objetivo):
    if nodo is None:
        return False
    if str(nodo.valor).lower() == str(objetivo).lower():
        return True
    return buscar_valor(nodo.izq, objetivo) or buscar_valor(nodo.der, objetivo)

print("Estructuras base cargadas correctamente.")
```

### Ejercicio 1: Árbol Genealógico (3 Niveles)

```python
# Nivel 1: Raíz
familia = Nodo("Abuela Carmen")

# Nivel 2: Hijos
familia.izq = Nodo("Tía Marta")
familia.der = Nodo("Papá Roberto")

# Nivel 3: Nietos
familia.izq.izq = Nodo("Primo Hugo")
familia.izq.der = Nodo("Prima Sara")
familia.der.izq = Nodo("Hermano Leo")
familia.der.der = Nodo("Yo")

print("--- Relaciones Familiares Directas ---")
print(f"Tronco principal: {familia.valor}")
print(f"Descendencia 1ra gen: {familia.izq.valor} y {familia.der.valor}")
print(f"Nietos por parte de {familia.izq.valor}: {familia.izq.izq.valor} y {familia.izq.der.valor}")
print(f"Nietos por parte de {familia.der.valor}: {familia.der.izq.valor} y {familia.der.der.valor}")
```

### Ejercicio 2: Árbol de Decisiones (Soporte Técnico de Hardware)

```python
soporte = Nodo("¿La PC enciende?")

# Respuestas Nivel 1
soporte.izq = Nodo("Sí: ¿Da video?")
soporte.der = Nodo("No: ¿Switch de la fuente en 'I'?")

# Acciones finales Nivel 2
soporte.izq.izq = Nodo("Sí: Sistema normal")
soporte.izq.der = Nodo("No: Reinsertar RAM DDR5")

soporte.der.izq = Nodo("Sí: Probar puente en placa")
soporte.der.der = Nodo("No: Reconectar cable")

print("--- Secuencia de Evaluación (Preorden) ---")
preorden(soporte)
print("\n\n--- Cierre de Diagnóstico (Postorden) ---")
postorden(soporte)
print()
```

### Ejercicio 3: Jerarquía de Archivos y Auditoría de Nodos

```python
server_dir = Nodo("Server_Minecraft_1.21")

server_dir.izq = Nodo("plugins")
server_dir.der = Nodo("world")

server_dir.izq.izq = Nodo("EliteMobs.jar")
server_dir.izq.der = Nodo("config.yml")

server_dir.der.izq = Nodo("playerdata")
server_dir.der.der = Nodo("level.dat")
server_dir.der.izq.izq = Nodo("uuid_cache.json") # Hoja profunda

print("--- Estadísticas del Directorio ---")
print(f"Total de elementos (Carpetas + Archivos): {contar_nodos(server_dir)}")
print(f"Archivos finales (Hojas): {contar_hojas(server_dir)}")

print("\n--- Búsqueda de Archivos ---")
archivo_buscar = input("Ingresa el nombre del archivo o carpeta a buscar: ")

if buscar_valor(server_dir, archivo_buscar):
    print(f"[OK] '{archivo_buscar}' fue encontrado en la jerarquía.")
else:
    print(f"[ERROR] '{archivo_buscar}' no existe en este directorio.")
```
