### Definición de Clases Base
*(Contiene todos los métodos solicitados para reutilizarlos en los 3 ejercicios)*

```python
class Nodo:
    def __init__(self, dato):
        self.dato = dato
        self.siguiente = None

class ListaEnlazada:
    def __init__(self):
        self.cabeza = None

    def insertar_inicio(self, dato):
        nuevo = Nodo(dato)
        nuevo.siguiente = self.cabeza
        self.cabeza = nuevo

    def insertar_final(self, dato):
        nuevo = Nodo(dato)
        if not self.cabeza:
            self.cabeza = nuevo
            return

        actual = self.cabeza
        while actual.siguiente:
            actual = actual.siguiente
        actual.siguiente = nuevo

    def recorrer(self):
        actual = self.cabeza
        if not actual:
            print("[Vacío]")
            return

        while actual:
            print(actual.dato, end=" -> ")
            actual = actual.siguiente
        print("Fin")

    def buscar(self, objetivo):
        actual = self.cabeza
        while actual:
            if actual.dato == objetivo:
                return True
            actual = actual.siguiente
        return False

    def eliminar(self, objetivo):
        actual = self.cabeza
        anterior = None

        while actual and actual.dato != objetivo:
            anterior = actual
            actual = actual.siguiente

        if not actual:
            return False # No se encontró

        if not anterior:
            self.cabeza = actual.siguiente
        else:
            anterior.siguiente = actual.siguiente
        return True

    def eliminar_inicio(self):
        if not self.cabeza:
            return None

        extraido = self.cabeza.dato
        self.cabeza = self.cabeza.siguiente
        return extraido

    def contar_elementos(self):
        contador = 0
        actual = self.cabeza
        while actual:
            contador += 1
            actual = actual.siguiente
        return contador

    def obtener_posicion(self, indice):
        if indice < 0:
            return "Error: Índice negativo no permitido."

        actual = self.cabeza
        for _ in range(indice):
            if not actual:
                return "Error: Índice fuera de los límites de la lista."
            actual = actual.siguiente

        if not actual:
            return "Error: Índice fuera de los límites de la lista."

        return actual.dato

print("Estructuras cargadas exitosamente.")
```

### Ejercicio 1: Historial de Navegación Simple

```python
print("--- Historial Web ---")
historial = ListaEnlazada()

# Añadimos URLs consecutivas simulando un navegador
historial.insertar_final("https://github.com/Oscxr29")
historial.insertar_final("https://campus.univo.edu.sv")
historial.insertar_final("https://stackoverflow.com/questions")
historial.insertar_final("https://youtube.com/watch")

print("Ruta de navegación:")
historial.recorrer()
```

### Ejercicio 2: Cola de Pendientes y Despacho de Tareas

```python
print("--- Gestor de Tareas ---")
pendientes = ListaEnlazada()

pendientes.insertar_final("Push al repo de algoritmos")
pendientes.insertar_final("Configurar servidor de Minecraft 1.21")
pendientes.insertar_final("Limpieza de hardware de la PC")

print("Lista inicial:")
pendientes.recorrer()

# Procesamos hasta que la cabeza quede en None
while pendientes.cabeza:
    tarea_actual = pendientes.eliminar_inicio()
    print(f"\n[✓] Completado: {tarea_actual}")
    print("Pendientes restantes: ", end="")
    pendientes.recorrer()
```

### Ejercicio 3: Filtro y Conteo de Nodos en Memoria

```python
print("--- Métricas del Sistema ---")
metricas = ListaEnlazada()

metricas.insertar_final("Uso CPU: 45%")
metricas.insertar_final("Uso RAM: 16GB")
metricas.insertar_final("Temp GPU: 65°C")
metricas.insertar_final("Disco C: 200GB libres")
metricas.insertar_final("Ping: 24ms")

total = metricas.contar_elementos()
print(f"Registros en memoria: {total}")

# Extraemos el índice 2 (que sería el 3er elemento)
indice = 2
valor_indice = metricas.obtener_posicion(indice)
print(f"Dato en la posición [{indice}]: {valor_indice}")
```
