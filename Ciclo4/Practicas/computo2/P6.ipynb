### Ejercicio 1: Mapeo de registros por código

```python
# Sistema de inventario
inventario_pc = {
    "COMP-01": {
        "nombre": "Intel Core i7-13700HX",
        "precio": 350.00,
        "stock": 5
    },
    "COMP-02": {
        "nombre": "NVIDIA RTX 4060",
        "precio": 299.99,
        "stock": 3
    },
    "COMP-03": {
        "nombre": "32GB RAM DDR5",
        "precio": 120.50,
        "stock": 10
    }
}

print("--- Inventario de Componentes ---")
for cod, detalle in inventario_pc.items():
    print(f"{cod} | {detalle['nombre']} | ${detalle['precio']} | Disp: {detalle['stock']}")

# Extraer un componente usando get para no dar error si no existe
busqueda = "COMP-02"
encontrado = inventario_pc.get(busqueda)

if encontrado:
    print(f"\nBúsqueda exitosa: {encontrado['nombre']} - {encontrado['stock']} unidades.")
else:
    print("\nEl componente no se encontró en la base de datos.")
```

### Ejercicio 2: Contador de ocurrencias en lista

```python
reportes = ["pantalla", "bateria", "teclado", "pantalla", "pantalla", "bateria", "limpieza", "teclado"]
frecuencia_reportes = {}

for rep in reportes:
    # Si no existe la clave, inicia en 0 y le suma 1
    frecuencia_reportes[rep] = frecuencia_reportes.get(rep, 0) + 1

print("Resumen de reparaciones solicitadas:")
for tipo, cant in frecuencia_reportes.items():
    print(f"- {tipo}: {cant} veces")
```

### Ejercicio 3: Agrupamiento por categorías

```python
lista_juegos = [
    ("Resident Evil 2", "Terror"),
    ("Minecraft", "Supervivencia"),
    ("GTA V", "Mundo Abierto"),
    ("Xonotic", "Shooter"),
    ("Rust", "Supervivencia")
]

juegos_categoria = {}

for titulo, genero in lista_juegos:
    if genero not in juegos_categoria:
        juegos_categoria[genero] = [] # Creamos la lista vacía si el género es nuevo
    juegos_categoria[genero].append(titulo)

print("Catálogo por Género:")
for genero, titulos in juegos_categoria.items():
    print(f"{genero}: {', '.join(titulos)}")
```
