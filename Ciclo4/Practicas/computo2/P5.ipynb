**Reto:** Simplificación y Operaciones con Lambda

Desarrollo de los 3 puntos solicitados: Refactorización, validaciones y procesamiento en lote.

```python
# 1. Refactorización de funciones convencionales a lambda

# Cada operación queda expresada en una sola línea con lambda.
calcular_total = lambda precio, descuento: round(precio * (1 - descuento / 100), 2)
convertir_celsius = lambda grados: (grados * 9 / 5) + 32
presentar_nombre = lambda nombre, apellido: f"{apellido.upper()}, {nombre.capitalize()}"

# Comprobación de resultados
print("Total después del descuento:", calcular_total(240, 12))
print("Temperatura en Fahrenheit:", convertir_celsius(30))
print("Nombre presentado:", presentar_nombre("Oscar", "Guevara"))
```

```python
# 2. Validaciones con operadores ternarios

# Clasificar un valor en positivo, negativo o cero.
clasificar_numero = lambda valor: "Positivo" if valor > 0 else ("Negativo" if valor < 0 else "Cero")

# Comprobar si una persona ya alcanzó la mayoría de edad.
es_adulto = lambda edad: edad >= 18

for valor in (7, -12, 0):
    print(f"Clasificación de {valor}:", clasificar_numero(valor))

print("¿Una persona de 19 años es adulta?:", es_adulto(19))
print("¿Una persona de 16 años es adulta?:", es_adulto(16))
```

```python
# 3. Procesamiento en lote usando lambda dentro de un bucle
montos_originales = [18.75, 42.00, 9.90, 86.40]
montos_actualizados = []

# Aplicar un incremento del 8 % a cada monto.
actualizar_monto = lambda monto: round(monto * 1.08, 2)

for monto in montos_originales:
    montos_actualizados.append(actualizar_monto(monto))

print("Montos originales:", [f"{monto:.2f}" for monto in montos_originales])
print("Montos con incremento:", [f"{monto:.2f}" for monto in montos_actualizados])
```
